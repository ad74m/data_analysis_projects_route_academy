# Task 1: Utility Functions
### Create the following functions:
- `add_numbers(a, b)`
- `subtract_numbers(a, b)`
- `multiply_numbers(a, b)`
- `divide_numbers(a, b)` (handle division by zero properly)

```python
def add_numbers(a, b):
    return a + b


def subtract_numbers(a, b):
    return a - b


def multiply_numbers(a, b):
    return a * b


def divide_numbers(a, b):
    if b == 0:
        return "Division by zero is not allowed"
    return a / b
```

### Test the functions

```python
print(add_numbers(5, 3))
print(subtract_numbers(5, 3))
print(multiply_numbers(5, 3))
print(divide_numbers(5, 3))
print(divide_numbers(5, 0))
```

# Task 2: Discount Calculator
### Create a function `apply_discount(price, discount_rate)`
- Use a lambda function to calculate the final price
- Ask for a price from the end user and return the discounted price

```python
def apply_discount(price, discount_rate):
    final_price = lambda p, d: p - (p * d / 100)
    return final_price(price, discount_rate)
```

### Ask the user for the price and the discount rate

```python
price = float(input("Enter the price: "))
discount_rate = float(input("Enter the discount rate: "))
print(f"Price Before Discount: {price}")
print(f"Discount Rate: {discount_rate}%")
print(f"Final Price: {apply_discount(price, discount_rate)}")
```

# Task 3: Console Calculator
### Build a calculator that:
- Uses functions only
- Supports addition, subtraction, multiplication and division
- Runs in a loop until the user chooses to exit
### Must include: input validation, a clear menu, reusable functions

```python
def add_numbers(a, b):
    return a + b


def subtract_numbers(a, b):
    return a - b


def multiply_numbers(a, b):
    return a * b


def divide_numbers(a, b):
    if b == 0:
        return "Division by zero is not allowed"
    return a / b


def menu():
    print("\n===== Calculator =====")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")


def operation(option, a, b):
    if option == 1:
        return add_numbers(a, b)
    elif option == 2:
        return subtract_numbers(a, b)
    elif option == 3:
        return multiply_numbers(a, b)
    elif option == 4:
        return divide_numbers(a, b)


def calculator():
    while True:
        menu()
        option = int(input("Choose an option (1-5): "))
        if option == 5:
            print("Exit")
            break
        a = int(input("Enter the first number: "))
        b = int(input("Enter the second number: "))
        print(f"Result: {operation(option, a, b)}")
```

### Run the calculator

```python
calculator()
```

### 2. Python Lists
https://www.hackerrank.com/challenges/python-lists/problem

```python
list = []
N = int(input())
for i in range(N):
    command = input().split()
    if command[0] == "insert":
        list.insert(int(command[1]), int(command[2]))
    elif command[0] == "print":
        print(list)
    elif command[0] == "remove":
        list.remove(int(command[1]))
    elif command[0] == "append":
        list.append(int(command[1]))
    elif command[0] == "sort":
        list.sort()
    elif command[0] == "pop":
        list.pop()
    elif command[0] == "reverse":
        list.reverse()
```

### 4. Dictionaries
https://www.hackerrank.com/challenges/30-dictionaries-and-maps/problem

```python
n = int(input())

phone_book = {}

for i in range(n):
    name, phone = input().split()
    phone_book[name] = phone

while True:
    name = input()
    if name in phone_book:
        print(name + "=" + phone_book[name])
    else:
        print("Not found")
        break
```

### 5. Nested Lists
https://www.hackerrank.com/challenges/nested-list/problem

```python
n = int(input())
students = []
for i in range(n):
    name = input()
    grade = float(input())
    students.append([name, grade])

grades = []
for student in students:
    if student[1] not in grades:
        grades.append(student[1])
grades.sort()

second_lowest = grades[1]
names = []
for student in students:
    if student[1] == second_lowest:
        names.append(student[0])

names.sort()
for name in names:
    print(name)
```

### 7. Mini-Max Sum
https://www.hackerrank.com/challenges/mini-max-sum/problem

```python
def miniMaxSum(arr):
    total = sum(arr)

    minimum = total - max(arr)
    maximum = total - min(arr)

    return f"{minimum} {maximum}"


numbers = []
for i in range(5):
    number = int(input())
    numbers.append(number)

print(miniMaxSum(numbers))
```
