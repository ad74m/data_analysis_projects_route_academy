### Ask the user to enter daily sales amounts until a negative number is entered.


### Add validation to handle non-numeric inputs gracefully.

### Store all valid sales values in a list.

```python
sales = []
while True:
    try:
        amount = float(input("Enter Daily Sales Amount: "))
        if amount < 0:
            break
        sales.append(amount)

    except ValueError:
        print("Please Enter a Numeric Value")
```

### Display Results

```python
print("Sales Amounts Entered:", sales)
```

### Create a dictionary where the key is the day number and the value is the sale amount.


```python
sales_dict = {}
for i in range(len(sales)):
    sales_dict[i + 1] = sales[i]

print("Sales Dictionary:", sales_dict)
```

### Calculate total sales, average sales, maximum and minimum sale.


### Create separate functions for each calculation.


```python
def calculate_total(sales):
    return sum(sales)


def calculate_average(sales):
    return sum(sales) / len(sales)


def calculate_maximum(sales):
    return max(sales)


def calculate_minimum(sales):
    return min(sales)


def display_sales(sales):
    print("Sales Amounts Entered:", sales)
    print("Total Sales Amount:", calculate_total(sales))
    print("Average Sales Amount:", calculate_average(sales))
    print("Maximum Sales Amount:", calculate_maximum(sales))
    print("Minimum Sales Amount:", calculate_minimum(sales))


display_sales(sales)
```
