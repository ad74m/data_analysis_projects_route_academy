# Part 2: HackerRank

## 1. Arrays
https://www.hackerrank.com/challenges/np-arrays/problem

```python
def reverse_array(arr):
    return np.array(arr[::-1], float)


arr = input().split()
reversed = reverse_array(arr)
print(reversed)
```

## 2. Shape and Reshape
https://www.hackerrank.com/challenges/np-shape-reshape/problem

```python
arr = input().split()
arr = np.array(arr, int)
arr = arr.reshape(3, 3)
print(arr)
```

## 3. Transpose and Flatten
https://www.hackerrank.com/challenges/np-transpose-and-flatten/problem

```python
N, M = input().split()
N, M = int(N), int(M)
arr = []
for i in range(N):
    arr.append(np.array(input().split(), int))

arr = np.array(arr)
print(np.transpose(arr))
print(arr.flatten())
```

## 4. Mean, Var, and Std
https://www.hackerrank.com/challenges/np-mean-var-and-std/problem

```python
N, M = input().split()
N, M = int(N), int(M)
arr = []
for i in range(N):
    arr.append(np.array(input().split(), int))

arr = np.array(arr)
print(np.mean(arr, axis=1))
print(np.var(arr, axis=0))
print(np.std(arr))
```

## 5. Sum and Product
https://www.hackerrank.com/challenges/np-sum-and-prod/problem

```python
N, M = input().split()
N, M = int(N), int(M)
arr = []
for i in range(N):
    arr.append(np.array(input().split(), int))

arr = np.array(arr)
print(np.prod(np.sum(arr, axis=0)))
```
